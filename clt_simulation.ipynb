# Objective:
## The primary goal of this project is to demonstrate the **Central Limit Theorem (CLT)** by simulating how random sample means transform into a **Gaussian Bell Curve.** By taking **20,000** iterations from a massive population of 1 million numbers, this simulation visually proves that as the sample size (k) increases from 4 to 50, the resulting distribution becomes significantly sharper and more precise. This transition highlights the core mathematical phenomenon where increasing sample sizes reduces variability, forcing the averages to converge tightly around the true population mean of 150 regardless of the initial data's shape.

# Methodology:
* **Population:** 1,000,000 integers generated randomly between 0 and 300
* **Sampling:** 20,000 independent samples were drawn
* **Comparing** Comparing the smoothness of the curve as the sample size increases

# Creating a dataset containing randomly generated 1 Million numbers

```python
import random
data_set = [random.randint(0,300) for _ in range(0,1000000)]
```

# Creating samples of randomly picked 4 numbers and creating a dataset of their respective means

```python
import numpy as np
means_4 = []
for i in range(0,20000):
    sample4  = random.sample(data_set,k=4)
    means_4.append(np.mean(sample4))
```

# Plotting Histogram means with 4

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(12, 4))
plt.hist(means_4,bins=100,color="steelblue")
plt.tight_layout()
plt.show()
```

# Creating samples of randomly picked 10 numbers and creating a dataset of their respective means

```python
means_10 = []
for i in range(0,20000):
    sample10  = random.sample(data_set,k=10)
    means_10.append(np.mean(sample10))
```

# Plotting histogram of means with 10

```python
plt.figure(figsize=(12, 4))
plt.hist(means_10,bins=100,color="steelblue")
plt.tight_layout()
plt.show()
```

# Creating samples of randomly picked 50 numbers and creating a dataset of their respective means

```python
means = []
for i in range(0,20000):
    sample50  = random.sample(data_set,k=50)
    means.append(np.mean(sample50))
```

# Plotting historgram of the means

```python
plt.figure(figsize=(12, 4))
plt.hist(means,bins=100,color="steelblue")
plt.tight_layout()
plt.show()
```

# Observation and Conclusion
* **Emergence of the Bell Curve:** Despite the raw data being a flat "Uniform" block, the distribution of the sample means transformed into a Bell Curve.
* **Order from Chaos:** This simulation proves that order exists even within massive randomness. No matter how messy the starting data is, the Bell Curve will eventually appear if the sample size is large enough.
* **Increment in Sample Size** As sample size increased the bell curve became smoother.
